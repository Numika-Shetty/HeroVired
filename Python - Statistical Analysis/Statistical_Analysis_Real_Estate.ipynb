**Statistics for Decision Making**

```python
import pandas as pd
from scipy import stats
```

```python
# Load the dataset
pd.set_option('display.max_columns', None) # Show all columns
df = pd.read_csv("C:\\Python\\01.dataset\\property.csv")
```

### # Question 1
#### ## For the suburb of Altona, it is postulated that a typical property sells for $800,000. Use the data at hand to test this assumption. Is the typical property price really $800,000 or has it increased? Use a significance level of 5%. 

```python

# Filter for suburb 'Altona'
altona = df[df['Suburb'].str.lower() == 'altona']

# Drop missing values
altona_prices = altona['Price'].dropna()

# Hypothesized population mean
population_mean = 800000

# Perform one-sample t-test (alternative: price > 800000)
t_stat, p_val = stats.ttest_1samp(altona_prices, population_mean)
p_val_one_tailed = p_val / 2 if t_stat > 0 else 1 - p_val / 2

print(f"T-statistic = {t_stat:.3f}, P-value (one-tailed) = {p_val_one_tailed:.4f}")

if p_val_one_tailed < 0.05:
    print("Reject H0: Property prices have increased beyond $800,000.")
else:
    print("Fail to reject H0: No significant increase beyond $800,000.")
```

### Question 2
#### For the year 2016, is there any difference in the prices of properties sold in the summer months vs winter months? Consider months from October till March as winter months and rest as summer months. Use a significance level of 5%.

```python
# --- Convert 'Date' column correctly (DD/MM/YYYY) ---
#df['Date'] = pd.to_datetime(df['Date'], format='%d/%m/%Y', errors='coerce')
df['Date'] = pd.to_datetime(df['Date'], dayfirst=True, errors='coerce')

# --- Filter data for the year 2016 and create a fresh copy ---
df_2016 = df[df['Date'].dt.year == 2016].copy()

# --- Extract the month from the date ---
df_2016.loc[:, 'Month'] = df_2016['Date'].dt.month

# --- Define Winter (Oct–Mar) and Summer (Apr–Sep) as per the question ---
winter = df_2016[df_2016['Month'].isin([10, 11, 12, 1, 2, 3])]
summer = df_2016[df_2016['Month'].isin([4, 5, 6, 7, 8, 9])]

# --- Perform an independent two-sample t-test (unequal variance) ---
t_stat, p_val = stats.ttest_ind(
    winter['Price'].dropna(),
    summer['Price'].dropna(),
    equal_var=False
)

# --- Display results ---
print(f"T-statistic = {t_stat:.3f}")
print(f"P-value = {p_val:.4f}")

if p_val < 0.05:
    print("Significant difference between Winter and Summer prices in 2016.")
else:
    print("No significant difference between Winter and Summer prices in 2016.")
```

### Question 3
For the suburb of Abbotsford, what is the probability that out of 10 properties sold, 3 will not have a car parking space? Use the column car in the dataset. Round off your answer to 3 decimal places. 

```python
# Filter for Abbotsford
abbotsford = df[df['Suburb'].str.lower() == 'abbotsford']

# Probability of not having car parking space (assuming 'car' column gives count)
p_no_car = (abbotsford['Car'].fillna(0) == 0).mean()

# Binomial probability: P(X=3) where X ~ Binomial(n=10, p=p_no_car)
n = 10
x = 3
prob_3_no_car = stats.binom.pmf(x, n, p_no_car)
print(f"P(X=3) = {prob_3_no_car:.3f}")
```
